## 1. Porțile sunt matrici

Starea $|0\rangle$ este vectorul $\begin{pmatrix}1\\0\end{pmatrix}$, iar $|1\rangle$ este $\begin{pmatrix}0\\1\end{pmatrix}$. O poartă este o matrice $2\times2$, iar aplicarea ei este o înmulțire matrice × vector.

$$I=\begin{pmatrix}1&0\\0&1\end{pmatrix}\quad X=\begin{pmatrix}0&1\\1&0\end{pmatrix}\quad Y=\begin{pmatrix}0&-i\\i&0\end{pmatrix}\quad Z=\begin{pmatrix}1&0\\0&-1\end{pmatrix}\quad H=\frac{1}{\sqrt2}\begin{pmatrix}1&1\\1&-1\end{pmatrix}$$

Dacă aplicăm întâi poarta $A$ și apoi poarta $B$, matricea totală este $B\cdot A$ (ordinea se inversează!).

În Python, `@` înseamnă înmulțire de matrici:

````python
I = np.array([[1, 0], [0, 1]])
X = np.array([[0, 1], [1, 0]])
Y = np.array([[0, -1j], [1j, 0]])
Z = np.array([[1, 0], [0, -1]])
H = np.array([[1, 1], [1, -1]]) / np.sqrt(2)

ket0 = np.array([1, 0])
print("X|0⟩ =", X @ ket0)
print("H|0⟩ =", np.round(H @ ket0, 3))
````

## 2. Faza — informația pe care nu o vedeți

Aplicați $Z$ pe $|+\rangle$: obțineți $|-\rangle = \frac{1}{\sqrt2}(|0\rangle - |1\rangle)$. Ambele stări dau **50% / 50%** la măsurare! Diferența este doar un semn minus — **faza**. Măsurarea directă nu o vede.

Truc: poarta H transformă faza în ceva vizibil. $H|+\rangle = |0\rangle$, iar $H|-\rangle = |1\rangle$.

````python
plus = QuantumCircuit(1)
plus.h(0)

minus = QuantumCircuit(1)
minus.h(0)
minus.z(0)

print("|+⟩ măsurat direct:", masoara(plus))
print("|-⟩ măsurat direct:", masoara(minus))

plus.h(0)
minus.h(0)
print("|+⟩ cu H înainte de măsurare:", masoara(plus))
print("|-⟩ cu H înainte de măsurare:", masoara(minus))
````
